```python
def one_away(first: str, second: str) -> bool:
    if len(first) == len(second):
        return replace(first, second)
    if len(first) + 1 == len(second):
        return insert(first, second)
    if len(first) - 1 == len(second):
        return insert(second, first)
    return False


def replace(first: str, second: str) -> bool:
    for i in range(len(first)):
        if first[i] != second[i]:
            return False
    return True


def insert(first: str, second: str) -> bool:
    index_1 = 0 
    index_2 = 0
    found_diff = False

    while index_1 < len(first) and index_2 < len(second):
        if first[index_1] != second[index_2]:
            if found_diff:
                return False
            else:
                index_2 += 1
        else:
            index_1 += 1
            index_2 += 1

        return True
    

print(one_away('pale', 'ple'))
print(one_away('pale', 'palebake'))
print(one_away('pale', 'bake'))
```

```python
from collections import Counter

# Palindrome Permutation
def is_pali_perm(input_str: str) -> bool:
    input_str = input_str.replace(' ', '').lower()
    counter = Counter(input_str)
    found_odd = False

    for v in counter.values():
        if v % 2 == 1:
            if found_odd:
                return False
            found_odd = True
    return True

print(is_pali_perm('Tact Coa'))
```

```python
# Implement an algorithm to determine if a string has all unique characters.
# What if you cannot use any data structures?

# careful many sorting algorithms take extra space
def isUnique(input: str) -> bool:
    arr = []
    for s in input:
        arr.append(s)
    arr.sort()
    for i in range(len(arr) - 1):
        if arr[i] == arr[i + 1]:
            return False
    return True

# With hash_map
def isUniqueMap(input: str) -> bool:
    map = {}
    for s in input:
        if s not in map:
            map[s] = 0
        else:
            return False
    return True

# Without any data structures O(n^2)
def isUnique2(input: str) -> bool:
    for i in input:
        count = 0
        for j in input:
            if i == j:
                count += 1
                if count > 1:
                    return False
    return True

# With bit vector
from bitarray import bitarray
def isUniqueBool(input: str) -> bool:
    # create an array of booleans of size 128 (number of ascii chars) that will indicate if the character is in the string
    # return False if you see a character again
    if len(input) > 128: # or 256 for extended ascii clarify with interviewer
        return False

    bool_arr = 128 * [False]
    for s in input:
        bit_val = ord(s)
        if bool_arr[bit_val]:
            return False
        bool_arr[bit_val] = True
    return True

input = 'asdf'
print(isUnique(input))
print(isUniqueMap(input))
print(isUnique2(input))
print(isUniqueBool(input))
```
