```python
def return_k_to_last(head: Node, k: int):   
    """
    iterative solution assuming length of linked list > 1
    assuming that k = 0 is the last element of the linked list
    start a counter at head, when counter == k create another pointer
    to that node
    Returns: k to last node or False if k out of bounds
    """    
    p1 = head
    p2 = False
    counter = 0

    while p1 is not None:
        if counter == k:
            p2 = head
        if counter > k:
            p2 = p2.next
        
        counter += 1
        p1 = p1.next

    return p2


def return_k_to_last2(head: Node, k: int) -> Node | bool:
    """
    Similar to the above solution but a bit cleaner
    Pros: time is O(n) and space O(1) because of the two pointers
    Cons: might be slightly harder to understand compared to the recursive solution
    """
    p1 = head
    p2 = head

    for i in range(k):
        if p1 is None:
            return False
        p1 = p1.next
    
    while p1 is not None:
        p1 = p1.next
        p2 = p2.next

    return p2


def return_k_to_last_recursive(head: Node, k: int):
    """
    recursive solution assuming k = 0 is the last node
    when we reach the end count num of steps from that place by 
    adding +1 to the previous function call
    
    Pros: less code
    Cons: uses O(n) memory where n is the length of the list 
    and we can't return the node itself
    """
    if head is None:
        return -1
    
    steps_from_end = 1 + return_k_to_last_recursive(head.next, k)
    if steps_from_end == k:
        print(head.data) # can't return the node to the call stack 
    
    return steps_from_end
    

# what if we need to return the node itself?
```

```python
class LinkedList:
    """
    Singly linked list implementation
    Part of the cracking the coding interview remove dups exercise
    Its main purpose is to test the solution. 
    Also supports addition and traversal 
    """
    def __init__(self):
        self.head = None
        self.tail = None
    

    def remove_dups(self) -> str:
        if self.head is None:
            return "List is empty"
        
        visited = set()
        cur = self.head
        prev = None

        while cur is not None:
            if cur.data in visited:
                prev.next = cur.next
            else:
                visited.add(cur.data)
                prev = cur

            cur = cur.next
        
        self.tail = prev

        return "Dups removed"


    def add(self, n: int) -> None:
        node = Node(n)
        if self.head is None:
            self.head = node
            self.tail = node
        else:
            self.tail.next = node
            self.tail = node


    def get_values(self) -> int | str:
        if self.head is None:
            return "List is empty"
        
        cur = self.head
        vals = []
        while cur is not None:
            vals.append(cur.data)
            cur = cur.next

        return vals


# Testing section
linked_list = LinkedList()
linked_list.get_values()
for i in range(10):
    n = random.randint(0, 5)
    linked_list.add(n)

with_dups = linked_list.get_values()
linked_list.remove_dups()
assert Counter(linked_list.get_values()) == Counter(set(with_dups))
```
