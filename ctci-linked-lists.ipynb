```python
import random
from collections import Counter

class LinkedList:
    """
    Sinlgy linked list implementation
    Part of the cracking the coding interview remove dups exercise
    Its main purpose is to test the solution. 
    Also supports addition and traversal 
    """
    def __init__(self):
        self.head = None
        self.tail = None
    

    def remove_dups(self) -> str:
        if self.head is None:
            return "List is empty"
        
        visited = set()
        cur = self.head
        prev = None

        while cur is not None:
            if cur.val in visited:
                prev.next = cur.next
            else:
                visited.add(cur.val)
                prev = cur

            cur = cur.next
        
        self.tail = prev

        return "Dups removed"


    def add(self, n: int) -> None:
        node = Node(n)
        if self.head is None:
            self.head = node
            self.tail = node
        else:
            self.tail.next = node
            self.tail = node


    def get_values(self) -> int | str:
        if self.head is None:
            return "List is empty"
        
        cur = self.head
        vals = []
        while cur is not None:
            vals.append(cur.val)
            cur = cur.next

        return vals


# Testing section
linked_list = LinkedList()
linked_list.get_values()
for i in range(10):
    n = random.randint(0, 5)
    linked_list.add(n)

with_dups = linked_list.get_values()
linked_list.remove_dups()
assert Counter(linked_list.get_values()) == Counter(set(with_dups))
```
