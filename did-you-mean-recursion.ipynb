```python
def rob(nums: List[int]) -> int:
    return rob_(nums, len(nums), {})


def rob_(nums, i, memo):
    if i == 0: # we don't steal any house
        return 0
    
    if i == 1: # we steal the first house
        return nums[0]

    if i in memo:
        return memo[i]

    # skip i house
    skip = rob_(nums, i - 1, memo)
    # steal i house
    steal = nums[i - 1] + rob_(nums, i - 2, memo)
    memo[i] = max(skip, steal) # store the result in memo

    return memo[i]   
```

```python
def jump_game_ii(nums: list[int]) -> int:
    return jump_game_ii_memo(0, nums, {})


def jump_game_ii_memo(cur: int, nums: list[int], cache: dict) -> int:
    """
    - recursively call the next step given the number of steps you can take from nums[cur]
    - if the next step is not in the end and falls at 0 continue 
    - if you are out of bounds return 0 steps from that position unless there is another way 
    to get to the end from that position that you found before
    - if can or have reached the end from the next position store the minimum n of steps
    you can take from cur
    - eventually cache[0] will bubble up with the 
    minimum number of steps to the end
    """
    if cur in cache:
        return cache[cur]
    if cur == len(nums) - 1:
        return 0
    if cur >= len(nums):
        return -1
    
    for i in range(1, nums[cur] + 1):
        step = cur + i
        # if next step is not the end and falls at 0 
        if step < len(nums) - 1 and nums[step] == 0:
            continue

        can_jump_steps = 1 + jump_game_ii_memo(step, nums, cache)
        if can_jump_steps == 0:
            if cur not in cache:
                return 0
            return cache[cur]
        else:
            if cur not in cache:
                cache[cur] = can_jump_steps
            else:
                cache[cur] = min(can_jump_steps, cache[cur])
    
    return cache[cur]


def jump_game_ii_iteration(nums: list[int]) -> int:
    """something with two pointers"""
    pass


nums = [2,3,1,1,4]
jump_game_ii(nums)
```

```python
def subsets(nums: List[int]) -> list[list[int]]:
    return subsets_(0, nums, [], [])


def subsets_(i, nums, cur_set, subs) -> list[list[int]]:
    if i == len(nums):
        subs.append(cur_set.copy())
        return

    cur_set.append(nums[i])
    subsets_(i + 1, nums, cur_set, subs)
    cur_set.pop()
    subsets_(i + 1, nums, cur_set, subs)
    return subs


nums = [1, 2, 3]
subsets(nums)
```

```python
def fibonacci_bottom_up(n: int) -> int:
    """
    fib(n) = fib(n - 2) + fib(n - 1)
    dp represents each operation above
    if n > 1 we start calculating each step by using dp[0] + dp[1] n - 1 times
    """
    if n < 2:
        return n

    i = 2
    dp = [0, 1] # [fib(0), fib(1)]
    while i <= n:
        temp = dp[1]
        dp[1] = dp[0] + temp
        dp[0] = temp
        i += 1

    return dp[1]


sequence = []
for n in range(9):
    sequence.append(fibonacci_bottom_up(n))
sequence
```
