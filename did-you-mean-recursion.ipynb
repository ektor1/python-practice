```python
def subsets(nums: List[int]) -> list[list[int]]:
    return subsets_(0, nums, [], [])


def subsets_(i, nums, cur_set, subs) -> list[list[int]]:
    if i == len(nums):
        subs.append(cur_set.copy())
        return

    cur_set.append(nums[i])
    subsets_(i + 1, nums, cur_set, subs)
    cur_set.pop()
    subsets_(i + 1, nums, cur_set, subs)
    return subs


nums = [1, 2, 3]
subsets(nums)
```

```python
def fibonacci(n: int, cache: dict) -> int:
    """fibonacci with caching"""
    if n in cache:
        return cache[n]

    if n == 1:
        return 1

    if n == 0:
        return 0

    cache[n] = fibonacci(n - 1, cache) + fibonacci(n - 2, cache)
    
    print(cache)
    return cache[n]


fibonacci(9, {})
```
