```python
def subsets(nums: List[int]) -> list[list[int]]:
    return subsets_(0, nums, [], [])


def subsets_(i, nums, cur_set, subs) -> list[list[int]]:
    if i == len(nums):
        subs.append(cur_set.copy())
        return

    cur_set.append(nums[i])
    subsets_(i + 1, nums, cur_set, subs)
    cur_set.pop()
    subsets_(i + 1, nums, cur_set, subs)
    return subs


nums = [1, 2, 3]
subsets(nums)
```

```python
def fibonacci_bottom_up(n: int) -> int:
    """
    fib(n) = fib(n - 1) + fib(n - 2)
    dp represents each operation above
    if n > 1 we start calculating each step by using dp[0] + dp[1] n times
    """
    if n < 2:
        return n

    i = 2
    dp = [0, 1] # [fib(0), fib(1)]
    while i <= n:
        temp = dp[1]
        dp[1] = dp[0] + temp
        dp[0] = temp
        i += 1

    return dp[1]


sequence = []
for n in range(9):
    sequence.append(fibonacci_bottom_up(n))
sequence
```

```python
def fibonacci(n: int, cache: dict) -> int:
    """fibonacci with caching"""
    if n in cache:
        return cache[n]

    if n == 1:
        return 1

    if n == 0:
        return 0

    cache[n] = fibonacci(n - 1, cache) + fibonacci(n - 2, cache)
    
    print(cache)
    return cache[n]


fibonacci(9, {})
```
