```python
def triple_step(stairs: int) -> int:
    """
    How many ways can you go up a staircase if you can do 1, 2, or 3 steps at a time
    Meaning: find number of paths from 0 to n given 1, 2, or 3 increments at a time
    """
    return triple_step_(stairs, 0)


def triple_step_(stairs: int, cur: int) -> int:
    ways = 0
    if cur == stairs:
        return 1
    if cur > stairs:
        return 0
    
    for i in range(1, 4):
        ways += triple_step_(stairs, cur + i)

    return ways 


triple_step_cache(3, 0, defaultdict(int))
```
