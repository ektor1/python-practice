```python
from collections import defaultdict


def triple_step(stairs: int) -> int:
    """
    How many ways can you go up a staircase if you can do 1, 2, or 3 steps at a time
    Meaning: find number of paths from 0 to n given 1, 2, or 3 increments at a time
    """
    res = triple_step_(stairs, 0)
    res_cache = triple_step_cache(stairs, 0, defaultdict(int))
    return f"res: {res}, res_cache: {res_cache}"


def triple_step_(stairs: int, cur: int) -> int:
    if cur == stairs:
        return 1
    if cur > stairs:
        return 0

    ways = 0
    for i in range(1, 4):
        ways += triple_step_(stairs, cur + i)

    return ways 


def triple_step_cache(stairs: int, cur: int, cache: dict) -> dict:
    if cur in cache:
        return cache[cur]
    if cur == stairs:
        return 1
    if cur > stairs:
        return 0

    ways = 0
    for i in range(1, 4):
        cache[cur] += triple_step_(stairs, cur + i) # triple_step_(stairs, cur + 1) + triple_step_(stairs, cur + 2) + triple_step_(stairs, cur + 3)

    return cache[cur] 

triple_step(9)
```
