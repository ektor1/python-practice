```python
def combinationSum(candidates, target):
    candidates.sort()
    combs = []
    combinationSum_(0, [], combs, candidates, target)
    return combs


def combinationSum_(start, cur_comb, combs, candidates, target):
    """
    passing i as the next start prevents us from including duplicate combinations in combs
    by ensuring that we don't start adding numbers we have already included in previous 
    paths. 
    """
    if target == 0:
        combs.append(cur_comb.copy())
        return
    
    for i in range(start, len(candidates)):
        num = candidates[i]
        if num > target:
            return
        cur_comb.append(num)
        combinationSum_(i, cur_comb, combs, candidates, target - num)
        cur_comb.pop()


def combinations_with_dupes(cur_comb, combs, candidates, target):
    if target == 0:
        combs.append(cur_comb.copy())
        return
    
    for num in candidates: # always starting form the beginning of candidates
        if num > target:
            return
        cur_comb.append(num)
        combinations_with_dupes(cur_comb, combs, candidates, target)
        cur_comb.pop()
```

```python
def subsets(self, nums: List[int]) -> List[List[int]]:
    res = []
    self.subsets_(res, [], 0, nums)
    return res    


def subsets_(self, res, cur_set, index, nums):
    if index == len(nums):
        res.append(cur_set.copy())
        return

    cur_set.append(nums[index])
    self.subsets_(res, cur_set, index + 1, nums)
    cur_set.pop()
    self.subsets_(res, cur_set, index + 1, nums)
```

```python
class TrieNode:
    def __init__(self):
        self.children = {} # {'char': TrieNode()}
        self.is_end = False


class Trie:

    def __init__(self):
        self.root = TrieNode()


    def insert(self, word: str) -> None:
        cur = self.root
        for char in word:
            if char not in cur.children:
                cur.children[char] = TrieNode()
            cur = cur.children[char]

        cur.is_end = True

    
    def delete(self, word: str):
        cur = self.root
        self.delete_(cur, word, 0)

    
    def delete_(self, node, word, index):
        if index == len(word):
            if not node.is_end:
                return False # word not found
            node.is_end = False

            # return to the parent if child node has children
            return not node.children

        next_char = word[index]
        next_node = node.children.get(next_char)
        if next_node is None:
            return False # word not found
        
        should_delete = self.delete_(next_node, word, index + 1)
        if should_delete:
            # delete child node if it has no children
            del node.children[next_char]

        return not node.children and not node.is_end


    def search(self, word: str) -> bool:
        cur = self.root
        for char in word:
            if char not in cur.children:
                return False
            cur = cur.children[char]
        
        return cur.is_end


    def startsWith(self, prefix: str) -> bool:
        cur = self.root
        for char in prefix:
            if char not in cur.children:
                return False
            cur = cur.children[char]
        
        return True
```

```python
from collections import defaultdict
def canFinish(numCourses: int, prerequisites: List[List[int]]) -> bool:
    adj_list = defaultdict(list)

    for course, pre in prerequisites:
        adj_list[course].append(pre)
    print(adj_list)

    for course in range(numCourses):
        if not is_cycle(course, adj_list, set()):
            return False

    return True


def is_cycle(cur, adj_list, seen):
    if cur in seen:
        return False

    if cur not in adj_list:
        return True

    seen.add(cur)
    for pre in adj_list[cur]:
        if not is_cycle(pre, adj_list, seen):
            return False
        
    seen.remove(cur) # backtrack
    del adj_list[cur] # avoid doing repeated work since we already know there are no cycles
    return True
```
