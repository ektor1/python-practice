```python
class Stack:
    """
    Implement a stack that keeps track of min value
    and can access it in O(1)
    store the min value at the current state of the stack in 
    a separate array. If we pop the current minimum, pop the last
    element from the min_val array
    """
    def __init__(self):
        self.stack = []
        self.min_val = [int('inf')]

    
    def push(self, val: int) -> None:
        self.stack.append(val)
        if val < self.min_val[-1]:
            self.min_val.append(val)
            

    def pop(self) -> int | str:
        if not self.stack:
            return "stack is empty"

        removed_val = self.stack.pop()
        if removed_val == self.min_val[-1]:
            self.min_val.pop()

        return removed_val


    def get_min_val(self) -> int | str:
        if not self.stack:
            return "stack is empty"
        return self.min_val[-1]
```
